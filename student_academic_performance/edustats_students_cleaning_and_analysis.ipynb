### genre

```python
df["Gender"].unique()
```

```python
for i in range(len(df)):
    if str(df.loc[i, "Gender"]).strip().upper().startswith("F"):
        df.loc[i, "Gender"] = "F"
    else:
        df.loc[i, "Gender"] = "M"
```

```python
df["Gender"].unique()
```

```python
df["Gender"].replace(to_replace=r'^\s*F.*', value='F', regex=True, inplace=True)
```

```python
df["Gender"].replace(to_replace=r'^\s*M.*', value='M', regex=True, inplace=True)
```

```python
df["Gender"].unique()
```

```python
df.tail(15)
```

```python
#unformisons la colonne "Gender" 
df["Gender"] = df["Gender"].apply(lambda x: "F" if str(x).strip().upper().startswith("F") else "M")
```

# ParentalEducation

```python
df["ParentalEducation"].unique()
```

```python
for i in range(len(df)):
    pe = str(df.loc[i, "ParentalEducation"]).strip().upper()
    if "H" in pe and "S" in pe:
        df.loc[i, "ParentalEducation"] = "High School"
    elif str(df.loc[i, "ParentalEducation"]).strip().upper().startswith("B") or str(df.loc[i, "ParentalEducation"]).strip().upper().startswith("L"):
        df.loc[i, "ParentalEducation"] = "Bachelor"
    elif "M" in pe and "S" in pe:
        df.loc[i, "ParentalEducation"] = "Master"
    else:
        df.loc[i, "ParentalEducation"] = np.nan

```

```python
df["ParentalEducation"].unique()
```
