### ANALYSE UNIVARIÉE

##### ANALYSE UNIVARIÉE : VARIABLES NUMÉRIQUES

```python
plt.figure(figsize=(6,4))
sns.histplot(df['age'], kde=True)
plt.title("Distribution de l'âge")

plt.show()
plt.figure(figsize=(6,3))
sns.boxplot(x=df['age'])
plt.title("Boxplot de l'âge")
plt.show()

print(df['age'].describe())
```

### Observation

**Âge entre 0 et ~82    Majorité entre 20 et 60**

```python
plt.figure(figsize=(6,4))
sns.histplot(df['avg_glucose_level'], kde=True)
plt.title("Distribution du glucose")
plt.show()

plt.figure(figsize=(6,3))
sns.boxplot(x=df['avg_glucose_level'])
plt.title("Boxplot du glucose")
plt.show()

print(df['avg_glucose_level'].describe())
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['bmi'], kde=True)
plt.title("Distribution du BMI")
plt.show()

plt.figure(figsize=(6,3))
sns.boxplot(x=df['bmi'])
plt.title("Boxplot du BMI")
plt.show()

print(df['bmi'].describe())
```

#### ANALYSE UNIVARIÉE : VARIABLES CATÉGORIELLES

```python
sns.countplot(data=df, x='gender')
plt.title("Répartition du genre")
plt.show()

tableau = pd.DataFrame({    'effectif': df['gender'].value_counts(),    'frequence': df['gender'].value_counts(normalize=True) * 100})
tableau
```

```python
sns.countplot(data=df, x='hypertension')
plt.title("Hypertension")
plt.show()

tableau_1 = pd.DataFrame({    'effectif': df['hypertension'].value_counts(),    'frequence': df['hypertension'].value_counts(normalize=True)*100})
tableau_1
```

```python
sns.countplot(data=df, x='smoking_status')
plt.title("Smoking status")
plt.xticks(rotation=45)
plt.show()
tableau_2 = pd.DataFrame({    'effectif': df['smoking_status'].value_counts(), 'frequence': df['smoking_status'].value_counts(normalize=True)*100})
tableau_2
```

### ANALYSE BIVARIÉE

```python
sns.boxplot(x=df['stroke'], y=df['age'])
plt.title("Age vs Stroke")
plt.show()
```

```python
sns.boxplot(x=df['stroke'], y=df['avg_glucose_level'])
plt.title("Glucose vs Stroke")
plt.show()
```

```python
sns.countplot(data=df, x='hypertension', hue='stroke')
plt.title("Hypertension vs Stroke")
plt.show()
```
