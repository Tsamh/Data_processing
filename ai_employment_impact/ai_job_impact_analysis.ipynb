# Analyse Exploratoire des Données

L'objectif est de comprendre comment l'adoption de l'IA transforme les conditions de travail, les salaires et les statuts d'emploi.

### 1.5 Variables les plus importantes

A partir de l'exploration initiale et du calcul des corrélations, cinq variables se distinguent comme particulièrement structurantes pour cette analyse.

**Job_Status - Statut d'emploi**
C'est la variable centrale de l'analyse. Elle indique si le poste est resté inchangé (*Unchanged*), a été modifié (*Modified*) ou supprimé (*Replaced*) suite à l'introduction de l'IA. C'est la variable qui résume le mieux l'impact de l'IA sur chaque employé.

**Automation_Risk - Niveau de risque d'automatisation**
Cette variable catégorielle (Low / Medium / High) mesure la vulnérabilité du poste à l'automatisation. Elle est directement liée au statut d'emploi et constitue le prédicteur le plus fort du remplacement par l'IA. Par exemple, un employé classé *High* a une probabilité bien plus élevée d'être déplacé.

**AI_Adoption_Level - Niveau d'adoption de l'IA dans l'entreprise**
Cette variable contextualise les impacts observés. Un employé évoluant dans une entreprise à forte adoption de l'IA est exposé à des transformations plus profondes qu'un employé dans une structure à adoption faible, même si leurs postes sont similaires.

**Productivity_Change_% - Variation de la productivité**
Variable numérique centrale. Elle mesure en pourcentage l'évolution de la productivité individuelle depuis l'introduction de l'IA. Sa distribution, ses corrélations avec l'expérience ou le secteur permettent d'évaluer si l'IA est réellement bénéfique pour la performance des employés.

**Salary_Before_AI / Salary_After_AI - Salaires avant et après l'IA**
Ces deux variables permettent de calculer l'impact économique direct de l'IA. Leur comparaison par secteur ou par niveau de risque révèle des disparités importantes : dans certains cas l'IA augmente les salaires (gain de valeur ajoutée), dans d'autres elle les fait baisser (remplacement partiel).

```python
df['Salary_Change'] = df['Salary_After_AI'] - df['Salary_Before_AI']
df['Salary_Change_Pct'] = ((df['Salary_After_AI'] - df['Salary_Before_AI']) / df['Salary_Before_AI'] * 100).round(2)

print('Deux variables derivees ont ete creees :')
print('  Salary_Change     : difference absolue de salaire (apres - avant)')
print('  Salary_Change_Pct : variation relative du salaire en pourcentage')

print('\nDistribution des variables cles :')
for col in ['Job_Status', 'Automation_Risk', 'AI_Adoption_Level']:
    print(f'\n{col} :')
    for k, v in df[col].value_counts().items():
        pct = v / len(df) * 100
        print(f'  {k:<22} {v:>5} employes   ({pct:.1f}%)')

print(f'\nProductivity_Change_% : moyenne = {df["Productivity_Change_%"].mean():.2f}%,  ecart-type = {df["Productivity_Change_%"].std():.2f}%')
print(f'Salary_Change_Pct    : moyenne = {df["Salary_Change_Pct"].mean():.2f}%,  ecart-type = {df["Salary_Change_Pct"].std():.2f}%')
```

```python
cols_corr = var_num + ['Salary_Change', 'Salary_Change_Pct']
corr_target = df[cols_corr].corr()['Productivity_Change_%'].drop('Productivity_Change_%').abs().sort_values(ascending=False)

print('Correlations (valeur absolue) avec Productivity_Change_% :')
for col, val in corr_target.items():
    barre = '#' * int(val * 30)
    print(f'  {col:<35} {val:.3f}  {barre}')
```

## Partie 2 - Datavisualisation avec Plotly

Cinq visualisations sont produites avec Plotly. Chacune apporte un éclairage complémentaire : la distribution des individus, les disparités de revenus, les relations entre variables, et l'impact sectoriel de l'IA.

### 2.1 Histogramme - Distribution de l'âge par niveau d'adoption de l'IA

Cet histogramme montre comment les employés de différentes tranches d'âge se répartissent selon le niveau d'adoption de l'IA dans leur entreprise. On cherche ici à savoir si l'âge est un facteur différenciant dans l'exposition à l'IA.

```python
fig_hist = px.histogram(
    df, x='Age', color='AI_Adoption_Level',
    barmode='overlay', nbins=25, opacity=0.72,
    color_discrete_sequence=['#EF4444', '#F59E0B', '#10B981'],
    title='Distribution de l age par niveau d adoption de l IA',
    labels={'Age': 'Age', 'AI_Adoption_Level': 'Adoption IA', 'count': 'Effectif'}
)
fig_hist.update_layout(
    paper_bgcolor='#0F172A', plot_bgcolor='#1E293B',
    font=dict(color='#E2E8F0', family='Arial'),
    margin=dict(l=60, r=40, t=60, b=50), title_font_size=16,
    legend=dict(bgcolor='#1E293B', bordercolor='#334155')
)
fig_hist.update_xaxes(gridcolor='#334155', color='#94A3B8', title='Age')
fig_hist.update_yaxes(gridcolor='#334155', color='#94A3B8', title='Effectif')
fig_hist.show()

print('Lecture : Les trois niveaux d adoption coexistent sur toutes les tranches d age.')
print('Les profils mid-career (35-55 ans) dominent dans les contextes d adoption elevee.')
```

### 2.2 Boxplot - Salaires avant et après l'IA par statut d'emploi

Ce boxplot compare la distribution des salaires avant et après l'introduction de l'IA, en les ventilant par statut d'emploi. Il permet de visualiser si les employés remplacés, modifiés ou inchangés présentaient des profils salariaux différents.

```python
df_melted = df.melt(
    id_vars=['Job_Status'],
    value_vars=['Salary_Before_AI', 'Salary_After_AI'],
    var_name='Periode', value_name='Salaire'
)
df_melted['Periode'] = df_melted['Periode'].map({
    'Salary_Before_AI': 'Avant IA',
    'Salary_After_AI': 'Apres IA'
})

fig_box = px.box(
    df_melted, x='Job_Status', y='Salaire', color='Periode', points=False,
    color_discrete_sequence=['#EF4444', '#10B981'],
    title='Distribution des salaires avant et apres IA par statut d emploi',
    labels={'Job_Status': 'Statut emploi', 'Salaire': 'Salaire ($)', 'Periode': 'Periode'}
)
fig_box.update_layout(
    paper_bgcolor='#0F172A', plot_bgcolor='#1E293B',
    font=dict(color='#E2E8F0', family='Arial'),
    margin=dict(l=60, r=40, t=60, b=50), title_font_size=16,
    legend=dict(bgcolor='#1E293B', bordercolor='#334155')
)
fig_box.update_xaxes(gridcolor='#334155', color='#94A3B8')
fig_box.update_yaxes(gridcolor='#334155', color='#94A3B8')
fig_box.show()

print('Lecture : Comparer les medianes et intervalles interquartiles par statut')
print('permet d identifier les groupes les plus penalises economiquement par l IA.')
```

### 2.3 Scatter plot - Expérience vs Changement de productivité

Ce nuage de points croise les années d'expérience avec le changement de productivité observé après l'IA. Chaque point représente un employé, coloré selon son niveau de risque d'automatisation et dimensionné selon ses heures de travail hebdomadaires.

```python
fig_scatter = px.scatter(
    df, x='Years_Experience', y='Productivity_Change_%',
    color='Automation_Risk', size='Work_Hours_Per_Week',
    hover_data=['Job_Role', 'Industry', 'Job_Status'],
    color_discrete_map={'Low': '#10B981', 'Medium': '#F59E0B', 'High': '#EF4444'},
    opacity=0.62,
    title='Experience professionnelle vs Changement de productivite (taille = heures/semaine)',
    labels={
        'Years_Experience': 'Annees d experience',
        'Productivity_Change_%': 'Changement de productivite (%)',
        'Automation_Risk': 'Risque d automatisation'
    }
)
fig_scatter.add_hline(
    y=0, line_dash='dash', line_color='#94A3B8',
    annotation_text='Seuil zero', annotation_font_color='#94A3B8'
)
fig_scatter.update_layout(
    paper_bgcolor='#0F172A', plot_bgcolor='#1E293B',
    font=dict(color='#E2E8F0', family='Arial'),
    margin=dict(l=60, r=40, t=60, b=50), title_font_size=16,
    legend=dict(bgcolor='#1E293B', bordercolor='#334155')
)
fig_scatter.update_xaxes(gridcolor='#334155', color='#94A3B8')
fig_scatter.update_yaxes(gridcolor='#334155', color='#94A3B8')
fig_scatter.show()

print('Lecture : L experience ne protege pas systematiquement contre une baisse de productivite.')
print('Le risque d automatisation est le facteur determinant de la position dans le graphique.')
```

### 2.4 Heatmap - Matrice de corrélation entre variables numériques

La matrice de corrélation mesure les associations linéaires entre toutes les variables numériques. Les valeurs proches de +1 ou -1 indiquent une relation forte ; les valeurs proches de 0 indiquent une relation faible ou absente. Cette visualisation permet aussi de détecter d'éventuels problèmes de multicolinéarité.

```python
corr_matrix = df[var_num].corr().round(2)

fig_heat = go.Figure(data=go.Heatmap(
    z=corr_matrix.values,
    x=corr_matrix.columns.tolist(),
    y=corr_matrix.columns.tolist(),
    colorscale='RdBu_r', zmid=0, zmin=-1, zmax=1,
    text=corr_matrix.values.round(2), texttemplate='%{text}',
    textfont=dict(size=11), hoverongaps=False,
    colorbar=dict(title='r', tickfont=dict(color='#E2E8F0'))
))
fig_heat.update_layout(
    title='Matrice de correlation - variables numeriques',
    paper_bgcolor='#0F172A', plot_bgcolor='#1E293B',
    font=dict(color='#E2E8F0', family='Arial'),
    margin=dict(l=60, r=40, t=60, b=60), title_font_size=16, height=500,
    xaxis=dict(color='#94A3B8', tickangle=-35),
    yaxis=dict(color='#94A3B8')
)
fig_heat.show()

print('Lecture : Les salaires avant et apres IA sont fortement correles (attendu).')
print('On note peu de multicolinearite problematique entre les autres variables numeriques.')
```

### 2.5 Graphique métier - Impact de l'IA par secteur et niveau de risque

Ce graphique à bulles est conçu pour répondre à la question métier centrale : quels secteurs sont les plus vulnérables à l'IA ? L'axe X représente la variation salariale moyenne, l'axe Y le taux de remplacement, et la taille des bulles l'effectif concerné. Les secteurs situés en haut à gauche sont les plus défavorisés.

```python
agg = (
    df.groupby(['Industry', 'Automation_Risk'])
    .agg(
        taux_remplacement=('Job_Status', lambda x: (x == 'Replaced').mean() * 100),
        variation_salaire=('Salary_Change_Pct', 'mean'),
        effectif=('Employee_ID', 'count'),
        productivite_moy=('Productivity_Change_%', 'mean')
    )
    .reset_index().round(2)
)

fig_metier = px.scatter(
    agg, x='variation_salaire', y='taux_remplacement',
    size='effectif', color='Automation_Risk', text='Industry',
    color_discrete_map={'Low': '#10B981', 'Medium': '#F59E0B', 'High': '#EF4444'},
    size_max=55,
    title='Impact de l IA par secteur et niveau de risque<br><sup>X = variation salariale (%)  |  Y = taux de remplacement (%)  |  Taille = effectif</sup>',
    labels={
        'variation_salaire': 'Variation salariale moyenne (%)',
        'taux_remplacement': 'Taux de remplacement (%)',
        'Automation_Risk': 'Risque d automatisation'
    },
    hover_data={'effectif': True, 'productivite_moy': ':.1f'}
)
fig_metier.update_traces(textposition='top center', textfont=dict(size=11))
fig_metier.add_vline(x=0, line_dash='dot', line_color='#94A3B8')
fig_metier.add_hline(
    y=agg['taux_remplacement'].mean(),
    line_dash='dot', line_color='#94A3B8',
    annotation_text='Taux moyen', annotation_font_color='#94A3B8'
)
fig_metier.update_layout(
    paper_bgcolor='#0F172A', plot_bgcolor='#1E293B',
    font=dict(color='#E2E8F0', family='Arial'),
    margin=dict(l=60, r=40, t=70, b=50), title_font_size=16, height=580,
    legend=dict(bgcolor='#1E293B', bordercolor='#334155')
)
fig_metier.update_xaxes(gridcolor='#334155', color='#94A3B8')
fig_metier.update_yaxes(gridcolor='#334155', color='#94A3B8')
fig_metier.show()

print('Lecture : Les secteurs en haut a gauche (fort taux de remplacement + baisse salariale)')
print('constituent la zone de vulnerabilite critique. L impact de l IA y est le plus defavorable.')
```

## Partie 3 - Dashboard interactif (Dash)

L'application Dash ci-dessous permet d'explorer les données de façon interactive. Elle reprend les cinq visualisations précédentes dans une interface unifiée, pilotée par trois filtres :

- **Secteur d'activité** (dropdown) - filtre tous les graphiques sur un secteur précis
- **Niveau de risque d'automatisation** (dropdown) - restreint l'analyse à un profil de risque
- **Tranche d'âge** (slider) - limite les données à une plage d'âge définie

Quatre indicateurs clés (KPI) se mettent à jour automatiquement avec les filtres :

- **Nombre d'employés dans la sélection** - effectif total après application des filtres
- **Taux de remplacement** - proportion d'employés dont le poste a été supprimé par l'IA
- **Productivité moyenne** - variation moyenne de la productivité dans la sélection
- **Variation salariale moyenne** - évolution moyenne des salaires en pourcentage

Pour lancer le dashboard, exécutez la cellule suivante puis ouvrez `http://127.0.0.1:8050` dans votre navigateur.

```python
from dash import Dash, dcc, html, Input, Output
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
import numpy as np

df = pd.read_csv('ai_job_impact.csv')
df['Salary_Change']     = df['Salary_After_AI'] - df['Salary_Before_AI']
df['Salary_Change_Pct'] = ((df['Salary_After_AI'] - df['Salary_Before_AI']) / df['Salary_Before_AI'] * 100).round(2)

INDUSTRIES   = sorted(df['Industry'].unique())
RISKS        = sorted(df['Automation_Risk'].unique())
AGE_MIN, AGE_MAX = int(df['Age'].min()), int(df['Age'].max())
NUM_COLS = ['Age', 'Years_Experience', 'Salary_Before_AI', 'Salary_After_AI',
            'Work_Hours_Per_Week', 'Job_Satisfaction', 'Productivity_Change_%']

BG = '#0F172A'; CARD = '#1E293B'; BORDER = '#334155'
TEXT = '#E2E8F0'; MUTED = '#94A3B8'
INDIGO = '#6366F1'; AMBER = '#F59E0B'; EMERALD = '#10B981'; RED = '#EF4444'

LAYOUT = dict(
    paper_bgcolor=CARD, plot_bgcolor=BG,
    font=dict(color=TEXT, family='Arial'),
    margin=dict(l=50, r=30, t=50, b=50), title_font_size=13,
    legend=dict(bgcolor=CARD, bordercolor=BORDER)
)
AXIS = dict(gridcolor='#334155', zerolinecolor='#475569', color=MUTED)

def card(children, style=None):
    base = {'background': CARD, 'border': f'1px solid {BORDER}',
            'borderRadius': '10px', 'padding': '14px',
            'boxShadow': '0 4px 20px rgba(0,0,0,.4)'}
    if style:
        base.update(style)
    return html.Div(children, style=base)

def kpi_card(label, value_id, color=INDIGO):
    return card([
        html.P(label, style={'color': MUTED, 'fontSize': '11px', 'margin': '0 0 4px'}),
        html.H3(id=value_id, style={'color': color, 'fontSize': '26px', 'margin': 0, 'fontWeight': 700})
    ], style={'flex': '1', 'minWidth': '150px', 'textAlign': 'center'})

app = Dash(__name__, suppress_callback_exceptions=True)
app.title = 'AI Job Impact Dashboard'

app.layout = html.Div(
    style={'background': BG, 'minHeight': '100vh', 'fontFamily': 'Arial, sans-serif', 'padding': '20px'},
    children=[
        card([html.H1('Impact IA sur Emplois - Dashboard Analytique',
                      style={'color': TEXT, 'margin': 0, 'fontSize': '22px', 'fontWeight': 800}),
              html.P('2 000 employes  |  17 variables  |  ai_job_impact.csv',
                     style={'color': MUTED, 'margin': '4px 0 0', 'fontSize': '12px'})],
             style={'marginBottom': '18px'}),

        card([html.Div([
            html.Div([html.Label('Secteur', style={'color': MUTED, 'fontSize': '11px', 'display': 'block', 'marginBottom': '4px'}),
                      dcc.Dropdown(id='dd-industry',
                                   options=[{'label': 'Tous les secteurs', 'value': 'ALL'}] + [{'label': i, 'value': i} for i in INDUSTRIES],
                                   value='ALL', clearable=False)],
                     style={'flex': '1', 'minWidth': '190px'}),
            html.Div([html.Label('Risque Automatisation', style={'color': MUTED, 'fontSize': '11px', 'display': 'block', 'marginBottom': '4px'}),
                      dcc.Dropdown(id='dd-risk',
                                   options=[{'label': 'Tous', 'value': 'ALL'}] + [{'label': r, 'value': r} for r in RISKS],
                                   value='ALL', clearable=False)],
                     style={'flex': '1', 'minWidth': '190px'}),
            html.Div([html.Label(id='slider-label', style={'color': MUTED, 'fontSize': '11px', 'display': 'block', 'marginBottom': '4px'}),
                      dcc.RangeSlider(id='slider-age', min=AGE_MIN, max=AGE_MAX,
                                      value=[AGE_MIN, AGE_MAX], step=1,
                                      marks={AGE_MIN: str(AGE_MIN), 40: '40', AGE_MAX: str(AGE_MAX)},
                                      tooltip={'placement': 'bottom', 'always_visible': True})],
                     style={'flex': '2', 'minWidth': '280px'}),
        ], style={'display': 'flex', 'gap': '20px', 'flexWrap': 'wrap', 'alignItems': 'flex-end'})],
             style={'marginBottom': '18px'}),

        html.Div([kpi_card('Employes filtres', 'kpi-count', INDIGO),
                  kpi_card('Taux de remplacement', 'kpi-replaced', RED),
                  kpi_card('Productivite moyenne', 'kpi-prod', EMERALD),
                  kpi_card('Variation salariale moy.', 'kpi-salary', AMBER)],
                 style={'display': 'flex', 'gap': '14px', 'marginBottom': '18px', 'flexWrap': 'wrap'}),

        html.Div([card([dcc.Graph(id='hist', config={'displayModeBar': False})], style={'flex': '1', 'minWidth': '360px'}),
                  card([dcc.Graph(id='box',  config={'displayModeBar': False})], style={'flex': '1', 'minWidth': '360px'})],
                 style={'display': 'flex', 'gap': '14px', 'marginBottom': '18px', 'flexWrap': 'wrap'}),

        html.Div([card([dcc.Graph(id='scatter', config={'displayModeBar': False})], style={'flex': '1', 'minWidth': '360px'}),
                  card([dcc.Graph(id='heatmap', config={'displayModeBar': False})], style={'flex': '1', 'minWidth': '360px'})],
                 style={'display': 'flex', 'gap': '14px', 'marginBottom': '18px', 'flexWrap': 'wrap'}),

        card([dcc.Graph(id='metier', config={'displayModeBar': False})], style={'marginBottom': '18px'}),

        html.P('Dashboard - Dash + Plotly  |  ai_job_impact.csv',
               style={'color': MUTED, 'textAlign': 'center', 'fontSize': '11px'})
    ]
)

@app.callback(
    Output('slider-label', 'children'), Output('kpi-count', 'children'),
    Output('kpi-replaced', 'children'), Output('kpi-prod', 'children'),
    Output('kpi-salary', 'children'),
    Output('hist', 'figure'), Output('box', 'figure'),
    Output('scatter', 'figure'), Output('heatmap', 'figure'), Output('metier', 'figure'),
    Input('dd-industry', 'value'), Input('dd-risk', 'value'), Input('slider-age', 'value'),
)
def update_all(industry, risk, age_range):
    dff = df.copy()
    if industry != 'ALL': dff = dff[dff['Industry'] == industry]
    if risk != 'ALL':     dff = dff[dff['Automation_Risk'] == risk]
    dff = dff[(dff['Age'] >= age_range[0]) & (dff['Age'] <= age_range[1])]

    slider_lbl = f'Age : {age_range[0]} - {age_range[1]} ans'
    kpi_count  = f'{len(dff):,}'
    kpi_rep    = f"{(dff['Job_Status'] == 'Replaced').mean() * 100:.1f} %"
    kpi_prod   = f"{dff['Productivity_Change_%'].mean():.1f} %"
    kpi_sal    = f"{dff['Salary_Change_Pct'].mean():.1f} %"

    fig_h = px.histogram(dff, x='Age', color='AI_Adoption_Level', barmode='overlay', nbins=20,
                         opacity=0.72, color_discrete_sequence=[RED, AMBER, EMERALD],
                         title='Distribution Age / Adoption IA',
                         labels={'Age': 'Age', 'count': 'Effectif'})
    fig_h.update_layout(**LAYOUT); fig_h.update_xaxes(**AXIS); fig_h.update_yaxes(**AXIS)

    dff_m = dff.melt(id_vars=['Job_Status'], value_vars=['Salary_Before_AI', 'Salary_After_AI'],
                     var_name='Periode', value_name='Salaire')
    dff_m['Periode'] = dff_m['Periode'].map({'Salary_Before_AI': 'Avant IA', 'Salary_After_AI': 'Apres IA'})
    fig_b = px.box(dff_m, x='Job_Status', y='Salaire', color='Periode', points=False,
                   color_discrete_sequence=[RED, EMERALD], title='Salaires par statut emploi',
                   labels={'Job_Status': 'Statut', 'Salaire': 'Salaire ($)'})
    fig_b.update_layout(**LAYOUT); fig_b.update_xaxes(**AXIS); fig_b.update_yaxes(**AXIS)

    fig_s = px.scatter(dff, x='Years_Experience', y='Productivity_Change_%',
                       color='Automation_Risk', size='Work_Hours_Per_Week', opacity=0.62,
                       color_discrete_map={'Low': EMERALD, 'Medium': AMBER, 'High': RED},
                       title='Experience vs Productivite',
                       labels={'Years_Experience': 'Experience (ans)', 'Productivity_Change_%': 'Productivite (%)'})
    fig_s.add_hline(y=0, line_dash='dash', line_color=MUTED)
    fig_s.update_layout(**LAYOUT); fig_s.update_xaxes(**AXIS); fig_s.update_yaxes(**AXIS)

    corr = dff[NUM_COLS].corr().round(2)
    short = ['Age', 'Exp', 'Sal.Avant', 'Sal.Apres', 'Heures', 'Satisf.', 'Prod%']
    fig_ht = go.Figure(go.Heatmap(
        z=corr.values, x=short, y=short, colorscale='RdBu_r', zmid=0, zmin=-1, zmax=1,
        text=corr.values.round(2), texttemplate='%{text}', textfont=dict(size=10), hoverongaps=False
    ))
    fig_ht.update_layout(**LAYOUT, title='Matrice de correlation',
                         xaxis=dict(color=MUTED, tickangle=-30), yaxis=dict(color=MUTED))

    if len(dff) < 5:
        fig_m = go.Figure().update_layout(**LAYOUT, title='Donnees insuffisantes pour ce filtre')
    else:
        agg = (dff.groupby(['Industry', 'Job_Status'])
               .agg(count=('Employee_ID', 'count'), avg_prod=('Productivity_Change_%', 'mean'),
                    avg_sal=('Salary_Change_Pct', 'mean'))
               .reset_index().round(2))
        fig_m = px.bar(agg, x='Industry', y='count', color='Job_Status', barmode='group',
                       color_discrete_map={'Unchanged': EMERALD, 'Modified': AMBER, 'Replaced': RED},
                       title='Statuts emploi par secteur',
                       labels={'count': "Nombre d employes", 'Industry': 'Secteur', 'Job_Status': 'Statut'})
        fig_m.update_layout(**LAYOUT); fig_m.update_xaxes(**AXIS); fig_m.update_yaxes(**AXIS)

    return slider_lbl, kpi_count, kpi_rep, kpi_prod, kpi_sal, fig_h, fig_b, fig_s, fig_ht, fig_m


import socket


def port_libre(depart=8050, essais=50):
    """Renvoie le premier port TCP libre a partir de `depart`.

    Le port n'est pas fige a 8050 : si une autre application Dash occupe
    deja ce port, Jupyter affiche l'iframe de cette application-la a la
    place de celle-ci.
    """
    for port in range(depart, depart + essais):
        with socket.socket() as sock:
            if sock.connect_ex(('127.0.0.1', port)) != 0:
                return port
    raise RuntimeError(
        'Aucun port libre entre %d et %d' % (depart, depart + essais))


PORT = port_libre()
print('Dashboard sur http://127.0.0.1:%d' % PORT)
app.run(debug=False, port=PORT, jupyter_mode='inline', jupyter_height=700)
```
